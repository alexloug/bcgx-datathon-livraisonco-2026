```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

DATA_DIR = Path("../data")

files = {
    "Champs-Elysées": "champs_elysees.csv",
    "Convention": "convention.csv",
    "Saint-Pères": "saints_peres.csv"
}

data = {}

for arc, filename in files.items():
    df = pd.read_csv(DATA_DIR / filename, sep=";")
    data[arc] = df
    print(f"{arc}: {df.shape}")
```

```python
for arc, df in data.items():
    print(f"\n===== {arc} =====")
    print("Dimensions:", df.shape)
    print("Colonnes:", df.columns.tolist())
    display(df.head())
```

```python
for arc, df in data.items():
    print(f"\n===== {arc} =====")
    print(df.isna().mean().mul(100).round(2))
```
